```python
fname = input("Enter first name: ").strip()
lname = input("Enter last name: ").strip()
sec = input("Enter section: ").strip()
subj = input("Enter favorite subject: ").strip()

fullname = fname + " " + lname

print("\nFormatted Output:")
print("Name:", fullname.title())
print("Name Code:", fullname.upper())
print("Section:", sec.upper())
print("Favorite Subject:", subj.title())
print("Character Count:", len(fullname))
```
