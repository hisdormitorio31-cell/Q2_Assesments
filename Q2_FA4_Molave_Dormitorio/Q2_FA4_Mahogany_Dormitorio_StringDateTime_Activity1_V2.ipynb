```python
# ask for student details
first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
favorite_subject = input("Enter favorite subject: ")

# combine first and last name and remove extra spaces
full_name = first_name.strip() + " " + last_name.strip()

# format the strings
proper_name = full_name.title()
name_code = full_name.upper()
clean_section = section.strip().upper()
clean_subject = favorite_subject.strip().title()
char_count = len(proper_name)

# display the output
print("\nFormatted Output:")
print("Name:", proper_name)
print("Name Code:", name_code)
print("Section:", clean_section)
print("Favorite Subject:", clean_subject)
print("Character Count:", char_count)
```
